# Salary vs. Years of Experience Analysis

## 1. Environment Setup & Data Import
Importing necessary libraries for data manipulation and visualization, then loading the dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Salary_dataset.csv")

# Preview first 5 rows
df.head()
```

## 2. Exploratory Data Analysis
Checking summary statistics like mean, standard deviation, and quartiles using `describe()`.

```python
# Display summary statistics
df.describe()
```

## 3. Data Visualization
Plotting the relationship between Years of Experience and Salary, and saving the figure for the README.

```python
# Set figure size
plt.figure(figsize=(8, 5))

# Create scatter plot
plt.scatter(df["YearsExperience"], df["Salary"], color="#1f77b4", alpha=0.8)

# Add titles and axis labels
plt.xlabel("Years Experience")
plt.ylabel("Salary")
plt.title("Salary vs Years Experience")

# Save image file locally for your GitHub README.md
plt.savefig("salary_vs_experience.png", dpi=300, bbox_inches="tight")

# Render plot
plt.show()
```
